```python
import numpy as np
```

```python
class MyLinear:
    def __init__(self, in_features, out_features):
        self.W = np.ones((in_features, out_features))
        self.b = np.ones((1, out_features))

    def forward(self, X):
        self.X = X
        return X @ self.W + self.b

    def backward(self, dL_dout, lr=0.01):
        dL_db = np.sum(dL_dout, axis=0, keepdims=True)
        dL_dW = self.X.T @ dL_dout
        dL_dX = dL_dout @ self.W.T

        self.b -= dL_db * lr
        self.W -= dL_dW * lr

        return dL_dX, dL_dW, dL_db

    def get_params(self):
        return self.W, self.b
```

```python
X = np.random.rand(50, 2)

true_W = np.array([[3], [5]])
true_b = 8

y = X @ true_W + true_b
```

```python
nn_layer = MyLinear(2, 1)

epochs = 1000

for i in range(epochs):
    y_pred = nn_layer.forward(X)
    a = y_pred - y
    b = a ** 2
    c = b.sum()
    loss = c / len(y)

    dL_dc = 1 / len(y)
    dL_db_loss = dL_dc * np.ones_like(b)
    dL_da = dL_db_loss * 2 * a 
    dL_dout = dL_da * 1

    dL_dX, dL_dW, dL_db = nn_layer.backward(dL_dout)

print('Веса модели: ', nn_layer.get_params())
print(true_W)
print(true_b)
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing()
X = data.data
y = data.target.reshape(-1, 1)
```

```python
def relu(x):
    return np.maximum(0, x)

def drelu(x):
    return (x > 0).astype(x.dtype)
```

```python
lin_1 = MyLinear(X.shape[1], 8)
lin_2 = MyLinear(8, 1)
```

```python
epochs = 1000

for i in range(epochs):
    z = lin_1.forward(X)
    y_pred = relu(z)
    y_pred = lin_2.forward(y_pred)

    a = y_pred - y
    b = a ** 2
    c = b.sum()
    loss = c / len(y)

    dL_dc = 1 / len(y)
    dL_db_loss = dL_dc * np.ones_like(b)
    dL_da = dL_db_loss * 2 * a 
    dL_dout = dL_da * 1

    dL_dX, dL_dW, dL_db = lin_2.backward(dL_dout, lr=0.001)
    dL_dX = dL_dX * drelu(z)
    dL_dX, dL_dW, dL_db = lin_1.backward(dL_dX, lr=0.001)


y_pred = lin_1.forward(X)
y_pred = relu(y_pred)
y_pred = lin_2.forward(y_pred)

print('RMSE: ', np.mean((y_pred - y) ** 2))
print('Веса модели: ', lin_1.get_params())
```
